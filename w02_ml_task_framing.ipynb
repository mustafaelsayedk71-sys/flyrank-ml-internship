## 1. My lane as an ML task (type)

*Classification, clustering, ranking, or scoring — which one, and why?*

```python
# ML Task Type: Ranking & Scoring / Regression
# Task Description: Predict an engagement/refresh potential score for content assets to rank which pieces of content should be prioritized for a refresh or optimization to maximize organic search discoverability and CTR.
```

## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

```python
# Target Variable: CTR_Opportunity_Score (or Proxy: impressions_90d * (1 - CTR)).
# Definition: Since directly predicting future CTR can be noisy, we construct a continuous target/proxy that measures underperforming high-impression opportunities. Higher values indicate content that gets seen often but clicked rarely, making it a high-priority candidate for a content refresh.
```

## 3. Success metric

*One metric you can defend. What number means 'good'?*

```python
# Primary Evaluation Metric: Mean Absolute Error (MAE) for continuous prediction, combined with NDCG@K (Normalized Discounted Cumulative Gain) for evaluating the quality of top-ranked content recommendations.
# Business Success Metric: A statistically significant lift in average CTR for content refreshed based on the model's top recommendations compared to a random baseline.
```

## 4. The unit of analysis, as a real dataframe

*Load your lane's slice and show it: one row = one what?*

```python
import pandas as pd
import numpy as np

# The dataset is already available locally after cloning the repository.
# Update data_path to point to the correct local file path.
data_path = 'flyrank-ml-internship/data/raw/content_refresh_anonymized.csv'

# 1. Load the dataset
df = pd.read_csv(data_path)

# 2. Unit of Analysis: One row = One specific URL / Content Piece
print(f"Dataset Shape: {df.shape}")
print(f"Unit of Analysis: Each row represents a unique content item.")

# 3. Construct/Sketch the Target Column (Proxy)
# Assuming 'CTR' column is actually named 'ctr' (lowercase) or an equivalent like 'engagement_rate'
# Let's try 'ctr' first, which is a common convention.
df['CTR_Opportunity_Score'] = df['impressions_90d'] * (1 - df['ctr'])

# Display sample output with the target column
df[['search_volume', 'impressions_90d', 'ctr', 'CTR_Opportunity_Score']].head()
```

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

```python
#A simple rule-based system (e.g., "refresh all URLs where CTR < 0.05") fails because it treats low-traffic pages and high-impression pages identically, leading to wasted resources.

#Machine Learning captures non-linear interactions across `search_volume`, `impressions_90d`, and `ctr` simultaneously. It accurately predicts potential upside to prioritize high-value optimization opportunities instead of applying rigid thresholds.
```
