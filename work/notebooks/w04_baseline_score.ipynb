## 1. My rule and its reason codes

*Write the rule in plain words first. Then the reason codes it can output.*

Rule in Plain Words: Calculate an Opportunity Score based on search visibility gap: $\text{Score} = \text{impressions\_90d} \times (1 - \text{ctr})$. High scores indicate pages with high search impressions but sub-optimal click rates, prioritizing them for content updates.Reason Codes Output:HIGH_IMPRESSION_LOW_CTR: Score $> 5000$ (Immediate refresh target).MODERATE_OPPORTUNITY: Score between $1000$ and $5000$ (Monitor & optimize).LOW_TRAFFIC_OR_HEALTHY_CTR: Score $< 1000$ (No action required).

## 2. Build the ranked queue (writes the CSV)

*Code the score, rank everything, write work/outputs/baseline_action_score.csv.*

```python
!git clone https://github.com/mustafaelsayedk71-sys/flyrank-ml-internship


%cd https://github.com/mustafaelsayedk71-sys/flyrank-ml-internship
```

```python
import pandas as pd
import numpy as np
import os

# 1. Load dataset
data_path = 'flyrank-ml-internship/data/raw/content_refresh_anonymized.csv'
df = pd.read_csv(data_path)

# 2. Calculate baseline opportunity score
df['baseline_score'] = df['impressions_90d'] * (1 - df['ctr'])

# 3. Assign action labels and reason codes
def assign_action(row):
    if row['baseline_score'] > 5000:
        return 'PRIORITY_REFRESH', 'HIGH_IMPRESSION_LOW_CTR'
    elif row['baseline_score'] > 1000:
        return 'MONITOR_OPTIMIZE', 'MODERATE_OPPORTUNITY'
    else:
        return 'NO_ACTION', 'LOW_TRAFFIC_OR_HEALTHY_CTR'

df[['action_label', 'reason_code']] = df.apply(assign_action, axis=1, result_type='expand')

# 4. Sort queue descending
ranked_queue = df.sort_values(by='baseline_score', ascending=False)

# 5. Save output CSV to work/outputs/
os.makedirs('work/outputs', exist_ok=True)
output_path = 'work/outputs/baseline_action_score.csv'
ranked_queue[['content_id', 'baseline_score', 'action_label', 'reason_code', 'impressions_90d', 'ctr']].to_csv(output_path, index=False)

print(f"Successfully generated ranked queue CSV at: {output_path}")
print("Top 5 Preview:")
print(ranked_queue[['content_id', 'baseline_score', 'action_label', 'reason_code']].head())
```
